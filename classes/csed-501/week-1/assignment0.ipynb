# More NumPy Practice

In the following section we will apply some of the more advanced NumPy concepts to a randomly generated array to show why NumPy operations are preferred to Python operations for speed.


To show these concepts, we'll first create some data to work with. We'll generate two arrays, $X \in \mathbb{R}^{n_x \times d}$ and $Y \in \mathbb{R}^{n_y \times d}$. These are NumPy arrays with shapes $(n_x, d)$ and $(n_y, d)$ where $n_x = 100, n_y = 150$, and $d = 64$.

$X$ and $Y$ represent points in $d$-dimensional space, where $X$ contains $n_x$ rows of $d$-dimensional points, and $Y$ contains $n_y$ rows of $d$-dimensional points. These points are drawn from a standard Normal distribution (mean=0, stdev=1) using [```np.random.standard_normal```](https://numpy.org/doc/2.1/reference/random/generated/numpy.random.standard_normal.html).



````python
import numpy as np

#Create a random number generator with a set random seed
#Setting the random seed enables reproducibility for comparing solutions; otherwise, everyone would have a different random array
rng = np.random.RandomState(seed=42)

#Set dimensions
nx, ny, d = 100, 150, 64

#Generate data (each entry is sampled from a standard normal distribution N(0, 1))
X = rng.standard_normal((nx, d))
Y = rng.standard_normal((ny, d))

#Print info about the arrays
#arr.shape provides the dimensions in a tuple for array arr
#arr.dtype provides the data type of the array arr
#arr.size provides the number of total elements in array arr
print(X.shape, X.dtype, X.size)
print(Y.shape, Y.dtype, Y.size)
````

5 **NumPy Operations vs Python Loops**

To highlight the importance of using NumPy instead of Python whenever possible and to practice some of the above concepts, you will implement functions which, given $X \in \mathbb{R}^{n_x \times d}$ and $Y \in \mathbb{R}^{n_y \times d}$, compute $D \in \mathbb{R}^{n_x \times n_y}$, the matrix of pairwise Euclidean distances between points in $X$ and $Y$. That is, $D$ is a NumPy array of shape $(n_x, n_y)$ with:

$$D[i, j] = \| X[i, :] - Y[j, :] \|_2 = \sqrt{\sum_{k=1}^d (X[i, k] - Y[j, k])^2} $$.

You will implement this three ways:

- Three nested Python loops (completely naive).
- Two nested loops (loop over points in $X$, loop over points in $Y$, vectorize only the distance computation).
- One Python loop (loop over points in $X$ and for each point in $X$ simultaneously vectorize all distances to points in $Y$).

We will also provide an example of a fourth even more efficient way to calculate this:
- Fully vectorized (use the property that for two points $x, y \in \mathbb{R}^{d}: \|x-y\|_2 = \sqrt{\|x\|_2 + \|y\|_2 - 2 x^\top y}$

Finally, we will compare the run time of the implementations.

5.1 Three Loops: Naive Python Implementation

````python
def euclidean_dist_three_loops(X, Y):

  #Get the dimensions of X and Y
  nx, d = X.shape   # syntax to conveniently retrieve elements from a tuple
  ny = Y.shape[0]

  #Create an array D to hold the pairwise euclidean distances
  #It is recommended to pre-allocate the result array when the shape and type is known ahead of time
  #because NumPy array handles both memory and access much better than pure python.
  D = np.empty((nx, ny), dtype=X.dtype)

  #Loop through each pair of d-dimensional points
  for i in range(nx):
    for j in range(ny):
      s = 0.0  # this will aggregate the sum of squared differences
      #Loop through each dimension of point X[i,:] and Y[j,:]
      for k in range(d):
        #TO-DO: compute diff = X[i, k] - Y[j, k]
        diff = None

        #TO-DO: add diff^2 to s

      #TO-DO: set D[i,j] to the square root of s (the euclidean distance between X[i,:] and Y[j,:])
      D[i, j] = None
  return D
````

5.2 Two Loops: Iterate over Pairs, Vectorize Distance Computation

````python
def euclidean_dist_two_loops(X, Y):

  #Get the number of rows (points) in X and Y
  nx = X.shape[0]
  ny = Y.shape[0]

  #Same as before, we pre-allocate the result array
  D = np.empty((nx, ny), dtype=X.dtype)

  #Loop through each pair of d-dimensional points
  for i in range(nx):
    for j in range(ny):
      #Now we take whole rows: X[i] is shape (d,), Y[j] is shape (d,)
      #In the previous version, we accessed coordinates explicitly: X[i, k], Y[j, k]
      #Now, NumPy computes the elementwise differences between the two entire rows at once.

      #TO-DO: compute diff, the difference between row i in X and row j in Y
      #This will produce an array of length d
      diff = None

      #TO-DO: square the differences, sum the squared differences, and place in D
      D[i,j] = None

  return D
````

5.3 One Loop: Iterate over Points in X, Vectorize Distance Computation to All Points in Y

````python
def euclidean_dist_one_loops(X, Y):

  #Get the number of rows (points) in X and Y
  nx = X.shape[0]
  ny = Y.shape[0]

  #Same as before, we pre-allocate the result array
  D = np.empty((nx, ny), dtype=X.dtype)

  #Loop through each row (point) in X
  for i in range(nx):

    #TO-DO: get the i-th point from X, shape (d, )
    x_pt = None

    #TO-DO: compute diff, the distance between x_pt and all points in Y
    #Numpy will automatically broadcast x_pt to (ny, d)
    #diff should have shape (ny, d)
    diff = None

    #TO-DO: square these distances

    #TO-DO: sum the squared differences using np.sum()
    #Note that we need to sum over the columns of diff as each row j in diff holds
    #the squared differences between the coordinates of x_pt and Y_j.
    #To do this, we use the `axis` parameter in np.sum to indicate which axis to sum over.
    # (axis=0 -> sum over rows -> output a vector of shape (n_cols, ))
    # (axis=1 -> sum over cols -> output a vector of shape (n_rows, ))
    # (if axis is not set or axis=None/(0, 1) -> sum over all entries -> output scalar)
    sum_sq = None # shape (ny, )

    #TO-DO: save the square root of sum_sq in D[i].
    #Note that we are simultaneously setting the distances on the whole row i,
    #representing distances between x_pt to all points in Y.
    D[i] = None

  return D
````

We also provide a fully vectorized, super quick implementation for comparison to the previous three methods. This function is fully implemented.

````python
def euclidean_dist_vectorized(X, Y):

  #Compute squared norms of each row
  #We do not do in place operations here because that would modify X and Y directly
  X_sq = np.sum(X ** 2, axis=1) # shape (nx,)
  Y_sq = np.sum(Y ** 2, axis=1) # shape (ny,)


  #Compute dot products between all X and Y
  # @ = matrix multiplication in NumPy, .T transposes the matrix.
  XY = X @ Y.T # shape (nx, ny)

  #Apply the formula: ||x_i - y_j||^2 = ||x_i||^2 + ||y_j||^2 - 2 x_i^T y_j
  # first, we reshape X_sq to shape (nx, 1). Now, if we sum X_sq with an array of shape (nx, ny), the i-th value
  # will be added to each row element.
  # -1 infers the length for that axis
  X_sq = np.reshape(X_sq, (-1, 1))
  # X_sq = X_sq[None, :] # equivalent syntax

  # similarly, we reshape Y_sq to shape (1, ny)(although NumPy handles this automatically)
  Y_sq = np.reshape(Y_sq, (1, -1))
  # Y_sq = Y_sq[:, None] # equivalent syntax

  # ||x_i - y_j||^2 = ||x_i||^2 + ||y_j||^2 - 2 x_i^T y_j
  D_sq = X_sq - 2 * XY + Y_sq

  # take element-wise square root.
  return np.sqrt(D_sq)
````
