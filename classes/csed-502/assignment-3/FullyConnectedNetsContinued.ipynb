```
import os
import sys

# Detect if we're running in Google Colab
try:
    import google.colab
    IN_COLAB = True
except ImportError:
    IN_COLAB = False

def setup_environment(assignment_folder='csed502/assignment3/'):
    """
    Set up the environment for running notebooks.

    Args:
        assignment_folder: Folder name in Google Drive (only used for Colab)

    Returns:
        bool: True if running in Colab, False if running locally
    """
    if IN_COLAB:
        print("Running in Google Colab")

        # Mount Google Drive
        from google.colab import drive
        drive.mount('/content/drive')

        # Verify folder name is set
        FOLDERNAME = assignment_folder
        if FOLDERNAME is None:
            raise ValueError("[!] Enter the foldername for Google Drive.")

        # Add to Python path
        sys.path.append('/content/drive/My Drive/{}'.format(FOLDERNAME))

        # Download CIFAR-10 dataset if needed
        os.chdir('/content/drive/My Drive/{}/csed502/datasets/'.format(FOLDERNAME))
        os.system('bash get_datasets.sh')
        os.chdir('/content/drive/My Drive/{}'.format(FOLDERNAME))

    else:
        print("Running locally")
        # Already in the correct directory when running locally

    return IN_COLAB

setup_environment(assignment_folder='csed502/assignment3/')
```
