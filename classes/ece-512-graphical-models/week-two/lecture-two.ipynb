```python
import networkx as nx
import matplotlib.pyplot as plt
import numpy as np
```

## Examples

### Directed Graph

```python
G_di = nx.gn_graph(16)
nx.draw(G_di, with_labels=True, font_weight='bold')
```

### Trees

```python
G_prefix_tree = nx.prefix_tree(["ab", "abs", "ad"])
nx.draw(G_prefix_tree, with_labels=True, font_weight='bold')
```

### Cliques

#### Ring of Cliques

```python
G_ring_of_cliques = nx.ring_of_cliques(3, 3)
nx.draw(G_ring_of_cliques, with_labels=True, font_weight='bold')
```

```python
print(f'Number of Cliques: {list(nx.enumerate_all_cliques(G_ring_of_cliques))}')
```

#### Erdős-Rényi graph or a binomial graph.

```python
G_rand = nx.fast_gnp_random_graph(16, 0.25, seed=42)
nx.draw(G_rand, with_labels=True, font_weight='bold')
```

```python
print(f'Number of Cliques: {list(nx.enumerate_all_cliques(G_rand))}')
print(f'Maximum Cliques: {list(nx.approximation.max_clique(G_rand))}')
print(f'Maximal Cliques: {list(nx.find_cliques(G_rand))}')
```

## Matrix Powers

#### Undirected Graph

```python
G_rand2 = nx.fast_gnp_random_graph(8, 0.35, seed=42)
nx.draw(G_rand2, with_labels=True, font_weight='bold')
```

```python
A = nx.to_numpy_array(G_rand2)
A
```

```python
np.linalg.matrix_power(A, 2)
```

```python
np.linalg.matrix_power(A, 3)
```

```python
np.linalg.matrix_power(A, 20)
```

```python
np.linalg.matrix_power(A, 50)
```

#### Directed
